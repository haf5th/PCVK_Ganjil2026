### Pertanyaan 2: Buat kanvas hitam berukuran sama dengan tipe data int16 dan int32. Bandingkan `dtype`, `min()`, `max()`, dan `nbytes` keduanya. Apakah tampilannya berbeda? Apakah ukuran memorinya berbeda? Jelaskan alasannya.

```python
# Kanvas hitam dengan int16
black_img_int16 = np.zeros((100, 100, 3), dtype=np.int16)
print('--- Kanvas int16 ---')
print(f'dtype: {black_img_int16.dtype}')
print(f'min(): {black_img_int16.min()}')
print(f'max(): {black_img_int16.max()}')
print(f'nbytes: {black_img_int16.nbytes} bytes')
plt.imshow(black_img_int16)
plt.title('Black Image (int16)')
plt.show()

# Kanvas hitam dengan int32
black_img_int32 = np.zeros((100, 100, 3), dtype=np.int32)
print('\n--- Kanvas int32 ---')
print(f'dtype: {black_img_int32.dtype}')
print(f'min(): {black_img_int32.min()}')
print(f'max(): {black_img_int32.max()}')
print(f'nbytes: {black_img_int32.nbytes} bytes')
plt.imshow(black_img_int32)
plt.title('Black Image (int32)')
plt.show()
```

#### Penjelasan Jawaban 2:
*   **`dtype`**: Ini menunjukkan tipe data elemen dalam array. `np.int16` menggunakan bilangan bulat 16-bit, sedangkan `np.int32` menggunakan bilangan bulat 32-bit.
*   **`min()` dan `max()`**: Untuk kanvas hitam yang dibuat dengan `np.zeros()`, semua nilai piksel adalah 0, sehingga `min()` dan `max()` keduanya akan menjadi 0. Perbedaan `dtype` tidak mempengaruhi nilai ini karena 0 berada dalam rentang kedua tipe data tersebut.
*   **`nbytes`**: Ini adalah jumlah total byte yang dikonsumsi oleh array dalam memori. `np.int32` membutuhkan lebih banyak memori per elemen (4 byte) dibandingkan `np.int16` (2 byte), sehingga `black_img_int32` akan memiliki `nbytes` yang dua kali lipat lebih besar dari `black_img_int16` untuk ukuran gambar yang sama.
*   **Tampilan**: Secara visual, kedua kanvas akan terlihat sama-sama hitam karena semua nilai pikselnya adalah 0, yang merepresentasikan warna hitam. Perbedaan `dtype` hanya berpengaruh pada rentang nilai yang bisa disimpan dan penggunaan memori, bukan pada tampilan jika semua nilai piksel adalah nol.
*   **Ukuran memori**: Ukuran memori berbeda karena `int32` menggunakan 32 bit (4 byte) per elemen, sedangkan `int16` menggunakan 16 bit (2 byte) per elemen. Karena jumlah piksel dan saluran warna sama, total byte untuk `int32` akan dua kali lipat lebih besar.

## TUGAS PRAKTIKUM - Analisa
Apakah mengubah ukuran citra menjadi lebih kecil dan kemudian mengembalikannya ke ukuran aslinya (resizing down dan up) akan mengubah rata-rata intensitas piksel pada citra tersebut? Jika ya, mengapa?

### Rancangan Program:
1.  **Muat Citra Asli**: Pastikan citra `img` (dalam format RGB) tersedia.
2.  **Hitung Rata-rata Intensitas Asli**: Hitung rata-rata semua nilai piksel pada citra asli.
3.  **Ubah Ukuran ke Lebih Kecil**: Resize citra asli ke ukuran yang lebih kecil (misal, setengah dari ukuran asli).
4.  **Ubah Ukuran Kembali ke Asli**: Resize citra yang lebih kecil kembali ke ukuran asli.
5.  **Hitung Rata-rata Intensitas Setelah Resizing**: Hitung rata-rata semua nilai piksel pada citra yang telah di-resize dua kali.
6.  **Bandingkan dan Tampilkan Hasil**: Cetak kedua rata-rata intensitas dan tampilkan citra asli serta citra yang telah di-resize dua kali untuk perbandingan visual.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
# 1. Hitung Rata-rata Intensitas Asli
# Hitung rata-rata intensitas dari citra asli
mean_intensity_original = np.mean(img)
print(f"Rata-rata intensitas piksel citra asli: {mean_intensity_original:.2f}")

# 2.Resize citra ke ukuran lebih kecil
original_height, original_width, _ = img.shape

# Tentukan faktor pengecilan (misal: 0.5 untuk setengah ukuran)
down_scale_factor = 0.5
new_width_down = int(original_width * down_scale_factor)
new_height_down = int(original_height * down_scale_factor)

img_down = cv.resize(img, (new_width_down, new_height_down), interpolation=cv.INTER_AREA)
print(f"Ukuran citra setelah dikecilkan: {img_down.shape}")

# --- 3. Resize citra kembali ke ukuran asli ---
img_up = cv.resize(img_down, (original_width, original_height), interpolation=cv.INTER_LINEAR)
print(f"Ukuran citra setelah dikembalikan ke ukuran asli: {img_up.shape}")

# --- 4. Hitung Rata-rata Intensitas Setelah Resizing ---
mean_intensity_resized = np.mean(img_up)
print(f"Rata-rata intensitas piksel citra setelah resize (down & up): {mean_intensity_resized:.2f}")

# --- 5. Bandingkan dan Tampilkan Hasil ---
plt.figure(figsize=(15, 7))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Citra Asli')

plt.subplot(1, 2, 2)
plt.imshow(img_up)
plt.title('Citra Setelah Resize (Down & Up)')

plt.suptitle('Perbandingan Citra Asli vs. Setelah Resizing (Down & Up)')
plt.show()

# Perbedaan intensitas
diff_intensity = mean_intensity_original - mean_intensity_resized
print(f"Perbedaan rata-rata intensitas: {diff_intensity:.2f}")
if abs(diff_intensity) > 0.01: # Ambil threshold kecil untuk perbedaan numerik
    print("\nKesimpulan: Rata-rata intensitas piksel berubah setelah operasi resize (mengecilkan lalu membesarkan kembali).")
else:
    print("\nKesimpulan: Rata-rata intensitas piksel tidak berubah signifikan setelah operasi resize (mengecilkan lalu membesarkan kembali).")
```

### Kesimpulan:
Dari data hasil di atas, terlihat bahwa rata-rata intensitas piksel citra **berubah** setelah operasi mengubah ukuran (mengecilkan lalu membesarkan kembali). Perubahan ini disebabkan oleh proses **interpolasi** yang digunakan oleh fungsi `cv.resize()`.

Ketika citra dikecilkan (`cv.INTER_AREA`), piksel-piksel akan digabungkan atau di-sampling, dan informasi detail dapat hilang. Ketika citra dibesarkan kembali (`cv.INTER_LINEAR`), piksel-piksel baru 'dibuat' berdasarkan rata-rata tertimbang dari piksel-piksel tetangga. Proses interpolasi ini (baik untuk mengecilkan maupun membesarkan) jarang menghasilkan nilai piksel yang identik dengan aslinya, karena ia mencoba untuk 'menebak' nilai piksel yang paling sesuai, yang dapat sedikit mengubah distribusi intensitas keseluruhan citra. Oleh karena itu, bahkan jika dimensi kembali ke ukuran semula, nilai-nilai piksel dan, akibatnya, rata-rata intensitasnya, kemungkinan besar akan sedikit berbeda.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

from google.colab import drive
drive.mount('/content/drive')
```

```python
def tambah_label(path_gambar, teks_label):
    """
    Membubuhkan bar label identitas pada bagian bawah citra.
    Semua ukuran (tinggi bar, skala font, ketebalan garis, posisi teks)
    dihitung secara proporsional dari dimensi citra masukan (img.shape),
    sehingga hasilnya konsisten secara visual pada citra ukuran berapa pun.
    """
    img = cv2.imread(path_gambar)
    if img is None:
        raise FileNotFoundError(f"File tidak ditemukan: {path_gambar}")

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = img_rgb.shape[:2]

    # 1) Tinggi bar label = 10% dari tinggi citra (proporsional, bukan piksel tetap)
    tinggi_bar = int(h * 0.10)

    # 2) Gambar bar solid (warna hitam) di bagian bawah citra
    cv2.rectangle(
        img_rgb,
        pt1=(0, h - tinggi_bar),
        pt2=(w, h),
        color=(0, 0, 0),
        thickness=-1   # -1 = solid/filled
    )

    # 3) Skala font dihitung relatif terhadap lebar citra
    #    (referensi: lebar 800px -> font_scale = 1.0)
    font_scale = w / 800

    # 4) Ketebalan teks proporsional terhadap lebar citra, minimal 1 px
    thickness = max(1, int(w / 400))

    # 5) Posisi teks dihitung relatif terhadap posisi bar & tinggi bar
    posisi_x = int(w * 0.02)
    posisi_y = h - int(tinggi_bar * 0.30)

    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(
        img_rgb,
        text=teks_label,
        org=(posisi_x, posisi_y),
        fontFace=font,
        fontScale=font_scale,
        color=(255, 255, 255),
        thickness=thickness,
        lineType=cv2.LINE_AA
    )

    # Cetak parameter yang dihitung otomatis (untuk pembuktian adaptivitas)
    print(f"[{path_gambar}]")
    print(f"  Ukuran citra   : {w} x {h}")
    print(f"  Tinggi bar     : {tinggi_bar} px  ({tinggi_bar/h*100:.1f}% dari tinggi)")
    print(f"  Font scale     : {font_scale:.3f}")
    print(f"  Thickness teks : {thickness} px")
    print(f"  Posisi teks    : ({posisi_x}, {posisi_y})")
    print("-" * 50)

    return img_rgb
```

```python
teks = "Dokumentasi Kegiatan PCVK 2026"

hasil_kecil = tambah_label("/content/drive/MyDrive/PCVK/WhatsApp Image 2026-10-07 at 11.27.35 AM.jpeg", teks)
hasil_besar = tambah_label("/content/drive/MyDrive/PCVK/ktm.jpeg", teks)

fig, axes = plt.subplots(1, 2, figsize=(14, 7))

axes[0].imshow(hasil_kecil)
axes[0].set_title(f"Foto Kecil — shape {hasil_kecil.shape}")

axes[1].imshow(hasil_besar)
axes[1].set_title(f"Foto Besar — shape {hasil_besar.shape}")

plt.suptitle("Label Identitas Adaptif pada Dua Ukuran Citra Berbeda", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
def buat_thumbnail_persegi(path_gambar, ukuran_kanvas=500,
                             warna_latar=(255, 255, 255), nama_toko="TOKO SAYA"):
    """
    Mengubah citra ukuran apapun (potrait/landscape/persegi) menjadi thumbnail
    persegi 1:1, dengan rasio asli tetap terjaga (tidak gepeng/melar), citra
    diposisikan tepat di tengah, dan logo nama toko berukuran proporsional.
    Semua perhitungan didasarkan pada ukuran_kanvas dan img.shape (tidak hardcode).
    """
    img = cv2.imread(path_gambar)
    if img is None:
        raise FileNotFoundError(f"File tidak ditemukan: {path_gambar}")

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = img_rgb.shape[:2]

    # 1) Hitung skala agar SISI TERPANJANG pas dengan ukuran_kanvas
    #    -> rasio asli (aspect ratio) tetap terjaga, tidak gepeng/melar
    skala = ukuran_kanvas / max(h, w)
    w_baru = int(w * skala)
    h_baru = int(h * skala)

    img_resized = cv2.resize(img_rgb, (w_baru, h_baru))

    # 2) Buat kanvas persegi dengan warna latar (padding/letterbox)
    kanvas = np.full((ukuran_kanvas, ukuran_kanvas, 3), warna_latar, dtype=np.uint8)

    # 3) Hitung offset agar citra tepat di TENGAH kanvas
    offset_x = (ukuran_kanvas - w_baru) // 2
    offset_y = (ukuran_kanvas - h_baru) // 2

    # 4) Tempelkan citra hasil resize ke kanvas menggunakan slicing NumPy
    kanvas[offset_y:offset_y + h_baru, offset_x:offset_x + w_baru] = img_resized

    # 5) Tambahkan logo/nama toko, ukuran & posisi proporsional terhadap ukuran_kanvas
    font = cv2.FONT_HERSHEY_DUPLEX
    font_scale = ukuran_kanvas / 700
    thickness = max(1, int(ukuran_kanvas / 300))

    posisi_x = int(ukuran_kanvas * 0.03)
    posisi_y = int(ukuran_kanvas * 0.06)

    cv2.putText(kanvas, text=nama_toko, org=(posisi_x, posisi_y),
                fontFace=font, fontScale=font_scale, color=(0, 0, 0),
                thickness=thickness, lineType=cv2.LINE_AA)

    # Cetak parameter untuk membuktikan adaptivitas & rasio yang terjaga
    print(f"[{path_gambar}]")
    print(f"  Ukuran asli     : {w} x {h}  (rasio {w/h:.3f})")
    print(f"  Ukuran resize   : {w_baru} x {h_baru}  (rasio {w_baru/h_baru:.3f})")
    print(f"  Skala           : {skala:.3f}")
    print(f"  Offset (x, y)   : ({offset_x}, {offset_y})")
    print(f"  Font scale logo : {font_scale:.3f}")
    print("-" * 55)

    return kanvas
```

```python
nama_toko = "TOKO SAYA"

thumb_potrait = buat_thumbnail_persegi(
    "/content/drive/MyDrive/PCVK/WhatsApp Image 2026-10-07 at 11.25.48 AM.jpeg",
    nama_toko=nama_toko
)

thumb_landscape = buat_thumbnail_persegi(
    "/content/drive/MyDrive/PCVK/WhatsApp Image 2026-10-07 at 11.27.35 AM.jpeg",
    nama_toko=nama_toko
)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(thumb_potrait)
axes[0].set_title("Thumbnail dari Foto Potrait")
axes[0].axis('off')

axes[1].imshow(thumb_landscape)
axes[1].set_title("Thumbnail dari Foto Landscape")
axes[1].axis('off')

plt.suptitle("Thumbnail Persegi 1:1 — Rasio Asli Tetap Terjaga", fontsize=14)
plt.tight_layout()
plt.show()
```
